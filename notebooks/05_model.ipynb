# Model Development

This notebook ingests `04_features_splits.joblib` which consists of all the train adn test splits for the various levels of apprenticeships. We will output a predicted and modelled output and apply a few modelling techniques to estimate the best classification. 

```python
import sys
from pathlib import Path
import joblib

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier,GradientBoostingClassifier

sys.path.append("..")
```

```python
splits = joblib.load(Path("../data/processed/04_features_splits.joblib"))

X_l3_train = splits['X_l3_train']
X_l3_test = splits['X_l3_test']
y_l3_train = splits['y_l3_train']
y_l3_test = splits['y_l3_test']

X_l4_train = splits['X_l4_train']
X_l4_test = splits['X_l4_test']
y_l4_train = splits['y_l4_train']
y_l4_test = splits['y_l4_test']

X_l7_train = splits['X_l7_train']
X_l7_test = splits['X_l7_test']
y_l7_train = splits['y_l7_train']
y_l7_test = splits['y_l7_test']
```

```python
models = {
    'LogisticRegression': LogisticRegression(random_state=42,class_weight='balanced'),
    'DecisionTree': DecisionTreeClassifier(random_state=42,class_weight='balanced'),
    'RandomForest': RandomForestClassifier(random_state=42,class_weight='balanced'),
    'GradientBoosting': GradientBoostingClassifier(random_state=42)
}

level_splits = {
    'L3': (X_l3_train, X_l3_test, y_l3_train, y_l3_test),
    'L4': (X_l4_train, X_l4_test, y_l4_train, y_l4_test),
    'L7': (X_l7_train, X_l7_test, y_l7_train, y_l7_test)
}

trained_models = {}
```

```python
for level,splits in level_splits.items():
    X_train,X_test,y_train,y_test = splits
    trained_models[level] = {}
    for model_name, model in models.items():
        print(f"Training {model_name} on {level}...")
        model.fit(X_train, y_train)
        trained_models[level][model_name] = model
        print(f"Done")
```

```python
joblib.dump(trained_models, Path("../data/processed/05_trained_models.joblib"))
```

### Final

Here I trained the various splits on 4 different models LogisticRegression and Decision Tree Classifiers to get a baseline and further on ensemble models to see if i could improve the performance using Random Forest and GradientBoosting. A nested loop was used to train all model and level combinations cleanly, avoiding repetitive code and making it easy to add new models or levels in future.With limited training data (58-109 rows per level), complex ensemble models risk overfitting — memorising training data rather than learning generalisable patterns. Model evaluation in 06_evaluate.ipynb will reveal whether this is the case. the `06_evaluate.ipynb` will run accuracy, precision and recall tests to check the performance of the various models on the various datasets.
